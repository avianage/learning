```python
# Install Python packages
%pip install anthropic python-dotenv

# Load env variable
from dotenv import load_dotenv
import os

load_dotenv()

# Create an API Client
from anthropic import Anthropic

client = Anthropic(
    api_key=os.getenv("ANTHROPIC_API_KEY"),
    base_url=os.getenv("ANTHROPIC_BASE_URI")
)
model = "claude-sonnet-4-6"

# Making a Helper Fxn
def add_user_message(messages, text):
    user_message = {"role":"user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role":"assistant", "content": text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model = model,
        max_tokens = 1000,
        messages = messages,
    )
    return message.content[0].text
```

```python
messages = []

add_user_message(messages,"How do i solve this problem 5x + 3 = 2 for x?" )
answer = chat(messages)

answer
```

Here, Claude is directly providing the solution, but if we want the user to think of a way to solve by giving hints, then we use a system prompt

```python
def chatSystemPrompt(messages, system = None):
    params = {
        "model":  model,
        "max_tokens":  1000,
        "messages":  messages,
    }

    if system:
        params["system"] = system

    message = client.messages.create(**params)

    return message.content[0].text
```

```python
messages1 = []
system = """
You are a patient math tutor.
Do not directly answer a student's questions.
Guide them to a solution step by step
"""
add_user_message(messages1,"How do i solve this problem 5x + 3 = 2 for x?" )
answer1 = chatSystemPrompt(messages1, system=system)

answer1
```
